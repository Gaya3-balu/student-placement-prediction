```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score
from sklearn.preprocessing import LabelEncoder
```

```python
df = pd.read_csv('/content/student_placement_synthetic.csv')
```

```python
df = df.drop('salary_package_lpa',axis=1)
```

```python
le = LabelEncoder()
df['branch'] = le.fit_transform(df['branch'])
df['college_tier'] = le.fit_transform(df['college_tier'])
```

```python
df.head(10)
```

```python
df.tail(10)
```

```python
df['placement_status'].value_counts()
```

```python
x = df.drop('placement_status',axis=1)
y = df['placement_status']
```

```python
x_train, x_test, y_train, y_test = train_test_split(x,y,test_size=0.2,stratify=y,random_state=42)
```

```python
model = LogisticRegression()
model.fit(x_train,y_train)
```

```python
x_train_predict = model.predict(x_train)
train_accuracy = accuracy_score(x_train_predict, y_train)
print(train_accuracy)
```

```python
x_test_predict = model.predict(x_test)
test_accuracy = accuracy_score(x_test_predict, y_test)
print(test_accuracy)
```
